```python
# Load images with dark, mid-range, and light tones
dark_tons = cv2.imread("/mnt/e/Image processing and computer vision using OpenCV/class.vision-master/images/dark.png")
mid_tons = cv2.imread("/mnt/e/Image processing and computer vision using OpenCV/class.vision-master/images/mid.png")
light_tons = cv2.imread("/mnt/e/Image processing and computer vision using OpenCV/class.vision-master/images/light.png")

# Convert the images from BGR to grayscale
dark_tons_gray = cv2.cvtColor(dark_tons, cv2.COLOR_BGR2GRAY)
mid_tons_gray = cv2.cvtColor(mid_tons, cv2.COLOR_BGR2GRAY)
light_tons_gray = cv2.cvtColor(light_tons, cv2.COLOR_BGR2GRAY)

# Display the original and grayscale images
images = [
    ("Dark Tones", dark_tons, dark_tons_gray),
    ("Mid Tones", mid_tons, mid_tons_gray),
    ("Light Tones", light_tons, light_tons_gray)
]

plt.figure(figsize=(12, 8))

for i, (title, original, gray) in enumerate(images):
    # Display the original image
    plt.subplot(3, 2, 2 * i + 1)
    plt.imshow(cv2.cvtColor(original, cv2.COLOR_BGR2RGB))
    plt.title(f"{title} - Original")
    plt.axis("off")

    # Display the grayscale image
    plt.subplot(3, 2, 2 * i + 2)
    plt.imshow(gray, cmap="gray")
    plt.title(f"{title} - Grayscale")
    plt.axis("off")

# Adjust the layout to prevent overlapping
plt.tight_layout()
plt.show()
```

```python
# Calculate the grayscale intensity histogram for the dark-toned image
dark_tons_gray_histogram = cv2.calcHist([dark_tons_gray], [0], None, [256], [0, 256])

# Calculate the grayscale intensity histogram for the mid-toned image
mid_tons_gray_histogram = cv2.calcHist([mid_tons_gray], [0], None, [256], [0, 256])

# Calculate the grayscale intensity histogram for the light-toned image
light_tons_gray_histogram = cv2.calcHist([light_tons_gray], [0], None, [256], [0, 256])

# Plot the grayscale histograms
plt.figure(figsize=(12, 6))

plt.plot(dark_tons_gray_histogram, color='black', label='Dark Tones')
plt.plot(mid_tons_gray_histogram, color='blue', label='Mid Tones')
plt.plot(light_tons_gray_histogram, color='orange', label='Light Tones')

# Set the histogram plot title and axis labels
plt.title('Grayscale Intensity Histograms')
plt.xlabel('Pixel Intensity')
plt.ylabel('Number of Pixels')

# Set the intensity range and display the legend
plt.xlim([0, 256])
plt.legend()
plt.grid(True, alpha=0.3)

# Display the histogram plot
plt.show()
```

```python
# Store the images and their titles
images = [
    (dark_tons, "Dark Tones"),
    (mid_tons, "Mid Tones"),
    (light_tons, "Light Tones")
]

# Define the colors of the BGR channels
channel_colors = ['blue', 'green', 'red']
channel_names = ['Blue', 'Green', 'Red']

# Plot the histogram of each channel for every image
for image, title in images:
    plt.figure(figsize=(10, 5))

    for channel, color, name in zip([0, 1, 2], channel_colors, channel_names):
        # Calculate the histogram for the current channel
        histogram = cv2.calcHist([image], [channel], None, [256], [0, 256])

        # Plot the channel histogram
        plt.plot(histogram, color=color, label=name)

    # Set the plot title and axis labels
    plt.title(f"{title} - Color Channel Histograms")
    plt.xlabel("Pixel Intensity")
    plt.ylabel("Number of Pixels")

    # Set the intensity range and display the legend
    plt.xlim([0, 256])
    plt.legend()
    plt.grid(alpha=0.3)

    # Display the histogram
    plt.show()
```
