```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
hue_spectrum = np.linspace(0, 179, num=180, dtype=np.uint8)
hue_spectrum
```

```python
image = np.array([hue_spectrum] * 30)
image.shape
```

```python
saturation_channel = np.ones_like(image, dtype=np.uint8) * 255
saturation_channel.shape
```

```python
value_channel = np.ones_like(image, dtype=np.uint8) * 255
value_channel.shape
```

```python
image_hsv = cv2.merge([image, saturation_channel, value_channel])
image_hsv.shape
```

```python
image_bgr = cv2.cvtColor(image_hsv, cv2.COLOR_HSV2BGR)
plt.imshow(image_bgr[..., ::-1])
```
