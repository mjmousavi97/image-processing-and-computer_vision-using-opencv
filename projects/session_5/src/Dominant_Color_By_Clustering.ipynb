```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
```

```python
image_path = "/mnt/e/Image processing and computer vision using OpenCV/class.vision-master/images/felfel-dolme.jpg"
image = cv2.imread(image_path)
print(image.shape)

plt.imshow(image[..., ::-1])
```

```python
flat_image = image.reshape((-1, 3))
print(flat_image.shape)
```

```python
n_cluster = 5

kmeans = KMeans(n_clusters=n_cluster)
kmeans.fit(flat_image)
```

```python
dominant_colors = np.array(kmeans.cluster_centers_, dtype=np.uint8)
dominant_colors
```

```python
labels = kmeans.labels_
labels
```

```python
kmeans.predict(flat_image[:10, :])
```

```python
np.unique(labels, return_counts=True)
```

```python
np.bincount(labels)
```

```python
plt.figure(figsize=(25, 5))

for i, color in enumerate(dominant_colors):

    image_zero = np.zeros((25, 25, 3), dtype=np.uint8)

    image_zero[:] = color

    plt.subplot(1, 5, i + 1)
    plt.imshow(image_zero[..., ::-1])
    plt.axis("off")

plt.show()
```
