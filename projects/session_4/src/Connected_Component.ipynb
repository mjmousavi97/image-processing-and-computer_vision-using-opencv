```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
image = cv2.imread("/mnt/e/Image processing and computer vision using OpenCV/class.vision-master/images/shapes.png", 0)
image.shape
```

```python
thresh, image = cv2.threshold(image, 127, 255, cv2.THRESH_BINARY)
```

```python
plt.imshow(image, cmap='gray')
```

```python
num_labels, labels = cv2.connectedComponents(image)
```

```python
num_labels
```

```python
labels.shape
```

```python
labeled_image = np.astype(labels * 179 / num_labels, np.uint8)
print(labeled_image.max())
print(labeled_image.min())
```

```python
blank = np.ones(shape=(labeled_image.shape[0], labeled_image.shape[1]))
blank = np.astype(blank*255, np.uint8)
print(blank.max())
print(blank.min())
```

```python
out_image = cv2.merge([labeled_image, blank, blank])
out_image = cv2.cvtColor(out_image, cv2.COLOR_HSV2BGR)

plt.imshow(out_image[..., ::-1])
```
