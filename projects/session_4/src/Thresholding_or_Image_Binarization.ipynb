```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = cv2.imread("/mnt/e/Image processing and computer vision using OpenCV/class.vision-master/images/gradient.png")
print(f"Shape of Image is: {image.shape}")

binarized = image.copy()
binarized[image >= 127] = 255
binarized[image < 127] = 0

plt.figure(figsize=(16, 8))
plt.subplot(1, 2, 1); plt.imshow(image[..., ::-1], cmap='gray'); plt.title("Original Image"); plt.axis("off")
plt.subplot(1, 2, 2); plt.imshow(binarized); plt.title("Binarized Image"); plt.axis("off");
```

```python
gradient_image = cv2.imread("/mnt/e/Image processing and computer vision using OpenCV/class.vision-master/images/gradient.jpg")

ret, threshold = cv2.threshold(src=gradient_image, thresh=127, maxval=255, type=cv2.THRESH_BINARY)

plt.figure(figsize=(16, 8))
plt.subplot(1, 2, 1)
plt.imshow(gradient_image[..., ::-1])

plt.subplot(1, 2, 2)
plt.imshow(threshold);
print(ret)
```

```python
image_plate = cv2.imread("/mnt/e/Image processing and computer vision using OpenCV/class.vision-master/images/plate.jpg", 0)

plt.figure(figsize=(16, 8))
plt.subplot(2, 3, 1); plt.imshow(image_plate, cmap='gray')
```

```python
import cv2
import matplotlib.pyplot as plt

ret1, thresh1 = cv2.threshold(image_plate, 127, 255, cv2.THRESH_BINARY)
ret2, thresh2 = cv2.threshold(image_plate, 127, 255, cv2.THRESH_BINARY_INV)
ret3, thresh3 = cv2.threshold(image_plate, 127, 255, cv2.THRESH_TRUNC)
ret4, thresh4 = cv2.threshold(image_plate, 127, 255, cv2.THRESH_TOZERO)
ret5, thresh5 = cv2.threshold(image_plate, 127, 255, cv2.THRESH_TOZERO_INV)

plt.figure(figsize=(16, 8))

plt.subplot(2, 3, 1); plt.imshow(image_plate, cmap='gray'); plt.title('Original')
plt.subplot(2, 3, 2); plt.imshow(thresh1, cmap='gray'); plt.title('BINARY')
plt.subplot(2, 3, 3); plt.imshow(thresh2, cmap='gray'); plt.title('BINARY_INV')
plt.subplot(2, 3, 4); plt.imshow(thresh3, cmap='gray'); plt.title('TRUNC')
plt.subplot(2, 3, 5); plt.imshow(thresh4, cmap='gray'); plt.title('TOZERO')
plt.subplot(2, 3, 6); plt.imshow(thresh5, cmap='gray'); plt.title('TOZERO_INV')

plt.tight_layout()
plt.show()
```

```python
global image_plate

def tracker(val):
    global th

    _, th = cv2.threshold(image_plate, val, 255, cv2.THRESH_BINARY)
    cv2.imshow('Threshold', th)

cv2.namedWindow("Threshold")
cv2.createTrackbar("Thrack", "Threshold", 0, 255, tracker)

tracker(0)

cv2.waitKey()
cv2.destroyAllWindows()

plt.imshow(th, cmap='gray')
```
