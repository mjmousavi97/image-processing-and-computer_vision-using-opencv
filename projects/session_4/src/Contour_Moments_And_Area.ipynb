```python
import numpy as np
import cv2
import matplotlib.pyplot as plt
```

```python
image = cv2.imread("/mnt/e/Image processing and computer vision using OpenCV/class.vision-master/images/shapes2.png")
print(f"Shape of image: {image.shape}")
plt.imshow(image[..., ::-1])
```

```python
image_binary = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
print(f"Shape of binary image: {image_binary.shape}")
plt.imshow(image_binary, cmap='gray')
```

```python
ret, thresh = cv2.threshold(image_binary, 250, 255, cv2.THRESH_BINARY_INV)
plt.imshow(thresh, cmap='gray')
```

```python
contours, hierarchy = cv2.findContours(thresh, cv2.RETR_TREE, cv2.CHAIN_APPROX_NONE)
print(len(contours))
```

```python
for contour in contours:
    M = cv2.moments(contour)
    cx = int(M['m10'] / M["m00"])
    cy = int(M['m01'] / M['m00'])
    cv2.circle(image, (cx, cy), 3, (255, 0, 0))

plt.imshow(image[..., ::-1])
```
