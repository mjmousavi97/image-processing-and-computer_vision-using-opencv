```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = cv2.imread("/mnt/e/Image processing and computer vision using OpenCV/class.vision-master/images/noisy.jpg", flags=0)
image.shape
```

```python
plt.imshow(image, cmap='gray')
```

```python
hist_image = cv2.calcHist([image], [0], None, [256], [0, 256])

plt.plot(hist_image)
plt.xlim([0, 256])
plt.show()
```

```python
ret1, thresh1 = cv2.threshold(image, 127, 255, cv2.THRESH_BINARY)

plt.imshow(thresh1, cmap='gray')
```

```python
ret2, thresh2 = cv2.threshold(image, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)

plt.imshow(thresh2, cmap='gray')
```

```python
blur_image = cv2.GaussianBlur(image, (5, 5), 0)

plt.imshow(blur_image, cmap='gray')
```

```python
hist_image_guassian = cv2.calcHist([blur_image], [0], None, [256], [0, 256])

plt.plot(hist_image_guassian)
plt.xlim([0, 256])
plt.show()
```

```python
ret3, thresh3 = cv2.threshold(blur_image, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)

plt.imshow(thresh3, cmap='gray')
```

```python
median_image = cv2.medianBlur(image, 5)

plt.imshow(median_image, cmap='gray')
```

```python
hist_image_median = cv2.calcHist([median_image], [0], None, [256], [0, 256])

plt.plot(hist_image_median)
plt.xlim([0, 256])
plt.show()
```

```python
ret4, thresh4 = cv2.threshold(median_image, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)

plt.imshow(thresh4, cmap='gray')
```
