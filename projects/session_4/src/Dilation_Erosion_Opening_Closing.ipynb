```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = cv2.imread("/mnt/e/Image processing and computer vision using OpenCV/class.vision-master/images/opencv_inv.png", 0)
plt.imshow(image, cmap='gray')
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

kernel = np.ones((5, 5), dtype=np.uint8)

erosion = cv2.erode(image, kernel=kernel, iterations=1)
dilation = cv2.dilate(image, kernel=kernel, iterations=1)
opening = cv2.morphologyEx(image, cv2.MORPH_OPEN, kernel=kernel)
closing = cv2.morphologyEx(image, cv2.MORPH_CLOSE, kernel=kernel)

plt.figure(figsize=(16, 8))

plt.subplot(1, 4, 1)
plt.imshow(erosion, cmap='gray')
plt.title("Erosion")
plt.axis('off')

plt.subplot(1, 4, 2)
plt.imshow(dilation, cmap='gray')
plt.title("Dilation")
plt.axis('off')

plt.subplot(1, 4, 3)
plt.imshow(opening, cmap='gray')
plt.title("Opening")
plt.axis('off')

plt.subplot(1, 4, 4)
plt.imshow(closing, cmap='gray')
plt.title("Closing")
plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
subtracted = cv2.subtract(dilation, erosion)

plt.imshow(subtracted, cmap='gray')
```
