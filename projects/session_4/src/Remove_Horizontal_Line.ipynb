```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = cv2.imread("/mnt/e/Image processing and computer vision using OpenCV/class.vision-master/images/notes.png", cv2.IMREAD_GRAYSCALE)
plt.imshow(image, cmap='gray')
```

```python
image_not = cv2.bitwise_not(image)
plt.imshow(image_not, cmap='gray')
```

```python
binary_image = cv2.adaptiveThreshold(image_not, 255, cv2.ADAPTIVE_THRESH_MEAN_C, cv2.THRESH_BINARY, 25, -2)
plt.imshow(binary_image, cmap='gray')
```

```python
cols = binary_image.shape[1]
kernel = cv2.getStructuringElement(cv2.MORPH_RECT, (cols// 30, 1))  #(width, height)
print(kernel)

open_image = cv2.morphologyEx(binary_image, cv2.MORPH_OPEN, kernel=kernel)
plt.imshow(open_image, cmap='gray')
```

```python
rows = binary_image.shape[0]
kernel_ver = cv2.getStructuringElement(cv2.MORPH_RECT, (1 , rows // 30))

open_vert = cv2.morphologyEx(binary_image, cv2.MORPH_OPEN, kernel=kernel_ver)
plt.imshow(open_vert, cmap='gray')
```

```python
kernel = cv2.getStructuringElement(cv2.MORPH_RECT, (2, 2))
print(kernel)
```

```python
dilate = cv2.dilate(open_vert, kernel=kernel)
plt.imshow(dilate, cmap='gray')
```
