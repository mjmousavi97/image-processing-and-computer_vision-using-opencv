```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image= cv2.imread("/mnt/e/Image processing and computer vision using OpenCV/class.vision-master/images/shapes.png", 0)
image.shape
```

```python
ret, image = cv2.threshold(image, 127, 255, cv2.THRESH_BINARY)
```

```python
contours, hierarchy = cv2.findContours(image, cv2.RETR_TREE, cv2.CHAIN_APPROX_SIMPLE)
```

```python
len(contours)
```

```python
for contour in contours:
    print(contour.shape)
```

```python
bgr_image= cv2.cvtColor(image, cv2.COLOR_GRAY2BGR)
```

```python
i = 0

cv2.polylines(bgr_image, [contours[i]], True, (0, 0, 255), 3)
plt.imshow(image[..., ::-1])
```
