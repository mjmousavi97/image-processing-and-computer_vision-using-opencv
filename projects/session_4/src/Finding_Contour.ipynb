```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = cv2.imread("/mnt/e/Image processing and computer vision using OpenCV/class.vision-master/images/rectangle.png", cv2.IMREAD_GRAYSCALE)
image.shape
```

```python
plt.imshow(image, cmap='gray')
```

```python
image.max()
```

```python
image.min()
```

```python
ret, thresh = cv2.threshold(image, 127, 255, cv2.THRESH_BINARY)
```

```python
plt.imshow(thresh, cmap='gray')
```

```python
thresh.max()
```

```python
contours, hierarchy = cv2.findContours(thresh, cv2.RETR_TREE, cv2.CHAIN_APPROX_NONE)
print(f"Number of contours: {len(contours)}")
```

```python
contours[0].shape
```

```python
image_bgr = cv2.cvtColor(image, cv2.COLOR_GRAY2BGR)
print(image_bgr.shape)
```

```python
for i in range(contours[0].shape[0]):
    point = contours[0][i][0]

    cv2.circle(image_bgr, point, 1, (255, 0, 0), 1) 

plt.imshow(image_bgr[..., ::-1])
```
