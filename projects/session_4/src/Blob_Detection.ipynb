```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = cv2.imread("/mnt/e/Image processing and computer vision using OpenCV/class.vision-master/images/blob.jpg")
print(image.shape)

plt.imshow(image[..., ::-1])
```

```python
# detect blobs and draw them on original image
detector = cv2.SimpleBlobDetector_create()

keypoints = detector.detect(image)

result1 = cv2.drawKeypoints(image, keypoints, np.array([]), (0, 0, 255))
result2 = cv2.drawKeypoints(image, keypoints, np.array([]), (255, 0, 0), cv2.DrawMatchesFlags_DRAW_RICH_KEYPOINTS)

plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1); plt.imshow(image[..., ::-1]); plt.title("Original")
plt.subplot(1, 3, 2); plt.imshow(result1[..., ::-1])
plt.subplot(1, 3, 3); plt.imshow(result2[..., ::-1])
```

```python
image2 = cv2.imread("/mnt/e/Image processing and computer vision using OpenCV/class.vision-master/images/blob_size.jpg")
print(image2.shape)

plt.imshow(image2[..., ::-1]);
```

```python
param = cv2.SimpleBlobDetector_Params()

param.filterByArea = True
param.minArea = 5000
param.maxArea = 20000

# detect blobs and draw them on original image
detector = cv2.SimpleBlobDetector_create(param)

keypoints = detector.detect(image2)

result1 = cv2.drawKeypoints(image2, keypoints, np.array([]), (0, 0, 255))
result2 = cv2.drawKeypoints(image2, keypoints, np.array([]), (255, 0, 0), cv2.DrawMatchesFlags_DRAW_RICH_KEYPOINTS)

plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1); plt.imshow(image2[..., ::-1]); plt.title("Original")
plt.subplot(1, 3, 2); plt.imshow(result1[..., ::-1])
plt.subplot(1, 3, 3); plt.imshow(result2[..., ::-1])
```
