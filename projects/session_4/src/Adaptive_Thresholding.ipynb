```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = cv2.imread("/mnt/e/Image processing and computer vision using OpenCV/class.vision-master/images/sudoku.jpg", 0)
plt.imshow(image, cmap='gray');
```

```python
ret1, thresh1 = cv2.threshold(image, 127, 255, cv2.THRESH_BINARY)
thresh2 = cv2.adaptiveThreshold(image, 255, cv2.ADAPTIVE_THRESH_MEAN_C, cv2.THRESH_BINARY, 11, 2)
thresh3 = cv2.adaptiveThreshold(image, 255, cv2.ADAPTIVE_THRESH_GAUSSIAN_C, cv2.THRESH_BINARY, 11, 2)

plt.figure(figsize=(16, 8))

plt.subplot(1, 4, 1)
plt.imshow(image, cmap='gray')
plt.title('Original Image')

plt.subplot(1, 4, 2)
plt.imshow(thresh1, cmap='gray')
plt.title('Global Thresholding (v=127)')

plt.subplot(1, 4, 3)
plt.imshow(thresh2, cmap='gray')
plt.title('Adaptive Mean Thresholding')

plt.subplot(1, 4, 4)
plt.imshow(thresh3, cmap='gray')
plt.title('Adaptive Gaussian Thresholding')

plt.tight_layout()
plt.show()
```

```python
median_denoised_image = cv2.medianBlur(image, 3)

ret1, thresh1 = cv2.threshold(median_denoised_image, 127, 255, cv2.THRESH_BINARY)
thresh2 = cv2.adaptiveThreshold(median_denoised_image, 255, cv2.ADAPTIVE_THRESH_MEAN_C, cv2.THRESH_BINARY, 11, 2)
thresh3 = cv2.adaptiveThreshold(median_denoised_image, 255, cv2.ADAPTIVE_THRESH_GAUSSIAN_C, cv2.THRESH_BINARY, 11, 2)

plt.figure(figsize=(16, 8))

plt.subplot(1, 4, 1)
plt.imshow(image, cmap='gray')
plt.title('Original Image')

plt.subplot(1, 4, 2)
plt.imshow(thresh1, cmap='gray')
plt.title('Global Thresholding (v=127)')

plt.subplot(1, 4, 3)
plt.imshow(thresh2, cmap='gray')
plt.title('Adaptive Mean Thresholding')

plt.subplot(1, 4, 4)
plt.imshow(thresh3, cmap='gray')
plt.title('Adaptive Gaussian Thresholding')

plt.tight_layout()
plt.show()
```
