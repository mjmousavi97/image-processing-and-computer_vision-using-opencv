```python
import cv2
import PIL
from PIL import ImageGrab
import numpy as np
```

```python
while True:
    screen = np.array(ImageGrab.grab())
    cv2.namedWindow("Screen")
    screen = cv2.cvtColor(screen, cv2.COLOR_RGB2BGR)
    cv2.imshow("Screen", screen)

    if cv2.waitKey(1) & 0xFF == 27:
        break

cv2.destroyAllWindows()
```
