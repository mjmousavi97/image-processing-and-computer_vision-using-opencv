```python
import cv2
import matplotlib.pyplot as plt
```

```python
cv2.namedWindow("WebCam")
cap = cv2.VideoCapture(0)
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))

fourcc = cv2.VideoWriter_fourcc(*'XVID')
out = cv2.VideoWriter("/mnt/e/Image processing and computer vision using OpenCV/output/video.avi", fourcc, 20, (height, width))

while cap.isOpened():
    ret, frame = cap.read()

    if not ret:
        print("Can't receive any frames! Please Check.")
        break

    frame = cv2.cvtColor(frame, code=cv2.COLOR_BGR2GRAY)
    out.write(frame)
    cv2.imshow("WebCam", frame)

    if cv2.waitKey(1) & 0xFF == 27:
        break

cap.release()
out.release()
cv2.destroyAllWindows()
    
```
