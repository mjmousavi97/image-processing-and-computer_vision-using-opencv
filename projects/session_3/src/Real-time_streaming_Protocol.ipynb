```python
import cv2
```

```python
cap = cv2.VideoCapture("rtsp://wowzaec2demo.streamlock.net/vod/mp4:BigBuckBunny_115k.mp4")

while True:
    ret, frame = cap.read()

    if ret != True:
        break
    cv2.namedWindow("RTSP")
    cv2.imshow("RTSP", frame)

    if cv2.waitKey(1) & 0xFF == 27:
        break

cap.release()
cv2.destroyAllWindows()
```
